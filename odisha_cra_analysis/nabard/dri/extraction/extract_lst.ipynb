```python
# =============================================================================
# MODIS LST EXTRACTION
# SEPTEMBER ONLY
# PIXELWISE RASTER EXPORT
# MOD11A2 LST
# 2000–2025
# =============================================================================

import ee
import geemap
import geopandas as gpd
from pathlib import Path

# =============================================================================
# INITIALIZE EARTH ENGINE
# =============================================================================

ee.Authenticate()

ee.Initialize(
    project='odishaextreme-heat'
)

# =============================================================================
# PATHS
# =============================================================================

BASE_DIR = Path.cwd()

geojson_path = BASE_DIR / "district.geojson"

# =============================================================================
# LOAD AOI
# =============================================================================

gdf = gpd.read_file(
    geojson_path
)

gdf["geometry"] = gdf.geometry.simplify(0.001)

aoi = geemap.gdf_to_ee(gdf)

# =============================================================================
# PARAMETERS
# =============================================================================

START_YEAR = 2000
END_YEAR = 2025

SCALE = 1000

# =============================================================================
# DATASET
# =============================================================================

modis_lst = ee.ImageCollection(
    'MODIS/061/MOD11A2'
)

# =============================================================================
# MONTHLY LST
# =============================================================================

def monthly_lst(year, month):

    start = ee.Date.fromYMD(
        year,
        month,
        1
    )

    end = start.advance(
        1,
        'month'
    )

    LST = (

        modis_lst

        .filterDate(start, end)

        .select('LST_Day_1km')

        .mean()

        .multiply(0.02)

        .subtract(273.15)

        .rename('LST')

    )

    return LST

# =============================================================================
# YEARWISE EXPORTS
# =============================================================================

for year in range(
    START_YEAR,
    END_YEAR + 1
):

    print(f"\nPreparing export for {year}")

    # =========================================================================
    # SEPTEMBER
    # =========================================================================

    LST_SEP = monthly_lst(
        year,
        9
    )

    LST_SEP = LST_SEP.rename(
        f'LST_SEP_{year}'
    )

    # =========================================================================
    # CLIP
    # =========================================================================

    LST_SEP = LST_SEP.clip(
        aoi.geometry()
    )

    # =========================================================================
    # EXPORT
    # =========================================================================

    task = ee.batch.Export.image.toDrive(

        image=LST_SEP,

        description=f'LST_SEP_{year}',

        folder='DRI_MODIS_LST',

        fileNamePrefix=f'LST_SEP_{year}',

        region=aoi.geometry(),

        scale=SCALE,

        maxPixels=1e13,

        fileFormat='GeoTIFF'

    )

    task.start()

    print(f"Export started for {year}")

# =============================================================================
# FINISHED
# =============================================================================

print("\nAll exports submitted.")
print("Open Earth Engine Tasks tab and click RUN.")
```
