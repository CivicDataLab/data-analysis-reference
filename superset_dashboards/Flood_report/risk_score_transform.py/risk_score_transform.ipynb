```python

def transform_dataset(df, rc_col='object-id', time_col='timeperiod', indicator_prefix=None, indicator_list=None):
    """
    Transforms a dataset by pivoting districts as columns and reshaping indicators.

    Parameters:
    df (pd.DataFrame): Input dataframe with districts, timeperiod, and indicators.
    district_col (str): Column name representing districts.
    time_col (str): Column name representing time periods.
    indicator_prefix (str, optional): Prefix to identify indicator columns.
    indicator_list (list, optional): List of indicator columns to consider.

    Returns:
    pd.DataFrame: Transformed dataframe with districts as columns and indicators as rows.
    """
    indicator_cols = indicator_list

    # Identify indicator columns based on prefix or provided list
    #if indicator_list:
    #    indicator_cols = indicator_list
    #elif indicator_prefix:
    #    indicator_cols = [col for col in df.columns if col.startswith(indicator_prefix)]
    #else:
        # Assume all columns except the district and timeperiod are indicators
    #    indicator_cols = [col for col in df.columns if col not in {district_col, time_col}]

    # Melt the DataFrame
    df_melted = df.melt(id_vars=[rc_col, time_col], value_vars=indicator_cols,
                         var_name='indicator', value_name='value')
    print(df_melted)
    # Pivot the DataFrame
    df_pivot = df_melted.pivot(index=[time_col, 'indicator'], columns=rc_col, values='value').reset_index()

    # Reorder columns (districts first, then timeperiod and indicator)
    district_cols = [col for col in df_pivot.columns if col not in {time_col, 'indicator'}]
    df_pivot = df_pivot[district_cols + [time_col, 'indicator']]

    return df_pivot
```
