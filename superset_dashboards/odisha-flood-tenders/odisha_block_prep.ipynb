### Dissolve village boundaries to blocks

Columns required: block_name, block_lgd, geometry, subdistrict, dtname

```python

# Select only the necessary columns
od_block = od[['dtname', 'dtcode11', 'block_name', 'block_lgd','geometry']]

# Dissolve based on the block column to merge village boundaries into blocks
od_block = od_block.dissolve(by='block_lgd', as_index=False)
```

```python
od_block.to_file('odisha_block.geojson')
```
